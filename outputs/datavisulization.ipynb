```python
import pandas as pd
from matplotlib.patches import Patch
from pathlib import Path


import matplotlib.pyplot as plt

df= Path(r"C:\Users\Alivinne\OneDrive\Desktop\workshop\student_performance_cleaned.csv")



df = pd.read_csv(df)
gender_count = (
    df["Gender"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({"m": "Male", "male": "Male", "f": "Female", "female": "Female"})
    .value_counts()
    .reindex(["Male", "Female"], fill_value=0)
)
plt.bar(gender_count.index,gender_count.values,color=['#ff99ff','#aa80ff'])
plt.title('Student Distribution')
plt.suptitle('Student Distribution by gender',fontsize=10,fontfamily='Arial',fontweight='bold',color='blue')
plt.xlabel('Gender',color='red')
plt.ylabel('No of students',color='violet')

for i, value in enumerate(gender_count.values):

    percentage = (value / gender_count.sum()) * 100

    plt.text(
        i,
        value + 20,
        f"{value} ({percentage:.0f}%)",
        ha="center",
        va="bottom",
    )
legend_elements = [
    Patch(facecolor='#ff99ff', label='MALE'),
    Patch(facecolor='#aa80ff', label='FEMALE')
]

plt.legend(
    handles=legend_elements,
    title='Gender',
    loc='upper right',
    fontsize=10,
)

plt.figtext(0.98,0.01,'Fig:1',ha='left',rotation=45)
```

```python
attendance_perf = df[["Attendance", "Final_Score", "Overall_Grade"]].dropna().copy()
attendance_perf["Overall_Grade"] = attendance_perf["Overall_Grade"].astype(str).str.upper().str.strip()

grade_order = ["A", "B", "C", "D", "E", "F"]
grade_colors = {
    "A": "#1f77b4",
    "B": "#2ca02c",
    "C": "#ff7f0e",
    "D": "#d62728",
    "E": "#9467bd",
    "F": "#8c564b",
}

attendance_perf["Overall_Grade"] = pd.Categorical(
    attendance_perf["Overall_Grade"], categories=grade_order, ordered=True
)

plt.figure(figsize=(10, 6))

for grade in grade_order:
    subset = attendance_perf[attendance_perf["Overall_Grade"] == grade]
    if subset.empty:
        continue
    plt.scatter(
        subset["Attendance"],
        subset["Final_Score"],
        s=60,
        alpha=0.7,
        color=grade_colors[grade],
        edgecolor="black",
        linewidth=0.5,
        label=grade
    )

corr = attendance_perf[["Attendance", "Final_Score"]].corr().iloc[0, 1]
plt.title("Attendance vs Academic Performance")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Score")
plt.legend(title="Overall Grade")
plt.grid(True, linestyle="--", alpha=0.4)

plt.annotate(
    f"Correlation: {corr:.2f}",
    xy=(0.02, 0.98),
    xycoords="axes fraction",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8)
)

plt.tight_layout()
import matplotlib.pyplot as _plt

class plt:
    def __init__(self):
        self._backend = _plt

    def __getattr__(self, name):
        return getattr(self._backend, name)

    def show(self, *args, **kwargs):
        return self._backend.show(*args, **kwargs)

    def savefig(self, *args, **kwargs):
        return self._backend.savefig(*args, **kwargs)

    def close(self, *args, **kwargs):
        return self._backend.close(*args, **kwargs)

    def tight_layout(self, *args, **kwargs):
        return self._backend.tight_layout(*args, **kwargs)

plt = plt()
plt.show()
```
