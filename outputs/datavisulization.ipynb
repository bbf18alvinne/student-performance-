```python
import pandas as pd
from matplotlib.patches import Patch
from pathlib import Path


import matplotlib.pyplot as plt

df= Path(r"C:\Users\Alivinne\OneDrive\Desktop\workshop\student_performance_cleaned.csv")



df = pd.read_csv(df)
gender_count = (
    df["Gender"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({"m": "Male", "male": "Male", "f": "Female", "female": "Female"})
    .value_counts()
    .reindex(["Male", "Female"], fill_value=0)
)
plt.bar(gender_count.index,gender_count.values,color=['#ff99ff','#aa80ff'])
plt.title('Student Distribution')
plt.suptitle('Student Distribution by gender',fontsize=10,fontfamily='Arial',fontweight='bold',color='blue')
plt.xlabel('Gender',color='red')
plt.ylabel('No of students',color='violet')

for i, value in enumerate(gender_count.values):

    percentage = (value / gender_count.sum()) * 100

    plt.text(
        i,
        value + 20,
        f"{value} ({percentage:.0f}%)",
        ha="center",
        va="bottom",
    )
legend_elements = [
    Patch(facecolor='#ff99ff', label='MALE'),
    Patch(facecolor='#aa80ff', label='FEMALE')
]

plt.legend(
    handles=legend_elements,
    title='Gender',
    loc='upper right',
    fontsize=10,
)

plt.figtext(0.98,0.01,'Fig:1',ha='left',rotation=45)
```

```python
plot_data = df[["School_Type", "Final_Score"]].copy()
plot_data["Final_Score"] = pd.to_numeric(plot_data["Final_Score"], errors="coerce")
plot_data["School_Type"] = plot_data["School_Type"].astype(str).str.strip()
plot_data = plot_data.dropna(subset=["Final_Score"])
plot_data = plot_data[plot_data["School_Type"] != ""]

school_types = sorted(plot_data["School_Type"].unique())
score_groups = [
    plot_data.loc[plot_data["School_Type"] == school, "Final_Score"].values
    for school in school_types
]

fig, ax = plt.subplots(figsize=(9, 6))

box = ax.boxplot(
    score_groups,
    patch_artist=True,
    showmeans=True,
    meanprops=dict(
        marker="D",
        markerfacecolor="#d62728",
        markeredgecolor="white",
        markersize=7
    ),
    medianprops=dict(color="#17202a", linewidth=2),
    whiskerprops=dict(color="#555555", linewidth=1.3),
    capprops=dict(color="#555555", linewidth=1.3),
    flierprops=dict(
        marker="o",
        markerfacecolor="#777777",
        markeredgecolor="white",
        markersize=5,
        alpha=0.7
    )
)

palette = ["#66c2a5", "#8da0cb", "#fc8d62", "#e78ac3"]
for patch, color in zip(box["boxes"], palette):
    patch.set_facecolor(color)
    patch.set_alpha(0.8)

ax.set_xticks(range(1, len(school_types) + 1))
ax.set_xticklabels([
    f"{school}\nn={len(scores)}"
    for school, scores in zip(school_types, score_groups)
])

for position, scores in enumerate(score_groups, start=1):
    median = pd.Series(scores).median()
    ax.annotate(
        f"Median: {median:.1f}",
        (position, median),
        xytext=(6, 5),
        textcoords="offset points",
        fontsize=9,
        color="#17202a"
    )

ax.scatter([], [], marker="D", color="#d62728", label="Mean")
ax.set_title(
    "Examination Scores by School Type",
    fontsize=14,
    fontweight="bold",
    color="#17202a",
    pad=14
)
ax.set_xlabel("School Type", fontsize=11)
ax.set_ylabel("Final Score", fontsize=11)
ax.set_ylim(0, max(100, plot_data["Final_Score"].max() + 10))
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.set_axisbelow(True)
ax.legend(frameon=False, loc="best")
plt.figtext(0.98, 0.1, 'Fig : 9.0', ha='right')

fig.tight_layout()
plt.show()
```
