```python
df = df.copy()

# Age: replace values outside [10, 25] with median of valid ages
valid_age_mask = df["Age"].between(10, 25)
age_median = df.loc[valid_age_mask, "Age"].median()
df["Age"] = df["Age"].where(valid_age_mask, age_median)

# Attendance: clamp to [0, 100]
df["Attendance"] = df["Attendance"].clip(lower=0, upper=100)

# Study_Hours: set negative values to 0
df["Study_Hours"] = df["Study_Hours"].clip(lower=0)

# Final_Score: clamp to [0, 100]
df["Final_Score"] = df["Final_Score"].clip(lower=0, upper=100)

df[["Age", "Attendance", "Study_Hours", "Final_Score"]].describe()


summary = df[["Age", "Attendance", "Study_Hours", "Final_Score"]].describe()
summary
```
